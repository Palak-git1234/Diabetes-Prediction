```python
#  PROJECT ON DIABETES PREDICTION 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# load
df = pd.read_csv('diabetes.csv')
print("=" * 45)
print("        DIABETES PREDICTION PROJECT")
print("=" * 45)
print("Total Patients  :", len(df))
print("Total Features  :", 8)
print("Diabetic        :", df['Outcome'].sum())
print("Healthy         :", df['Outcome'].value_counts()[0])   # 0=non diabetes, 1=diabetes
print("=" * 45)

# preprocessing
cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
for col in cols:
    df[col] = df[col].replace(0, df[col].mean())

x = df.drop('Outcome', axis=1)
y = df['Outcome']

# split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# scale
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)

print("\n                DATA SPLIT")
print("-" * 45)
print("Training Samples :", len(x_train))
print("Testing Samples  :", len(x_test))

# ---------LOGISTIC REGRESSION MODEL---------
model = LogisticRegression()
model.fit(x_train, y_train)
print("\nModel Training   : DONE.")

y_pred = model.predict(x_test)
acc = accuracy_score(y_test, y_pred) * 100

print("\n" + "=" * 45)
print("        LOGISTIC REGRESSION RESULTS")
print("=" * 45)
print(" Accuracy :", round(acc, 2), "%")

cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap="Blues")
plt.title('Confusion Matrix - Logistic Regression')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

report = classification_report(y_test, y_pred, output_dict=True)
print("-" * 45)
print("\n              DETAILED REPORT")
print("-" * 45)
print("No Diabetes:")
print("Precision :", round(report['0']['precision'] * 100, 2), "%")
print("Recall    :", round(report['0']['recall']    * 100, 2), "%")
print("F1 Score  :", round(report['0']['f1-score']  * 100, 2), "%")
print("\nDiabetes:")
print("Precision :", round(report['1']['precision'] * 100, 2), "%")
print("Recall    :", round(report['1']['recall']    * 100, 2), "%")
print("F1 Score  :", round(report['1']['f1-score']  * 100, 2), "%")
print("=" * 45)

# ------------DECISION TREE MODEL-------------

dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(x_train, y_train)
dt_pred = dt_model.predict(x_test)
dt_acc = accuracy_score(y_test, dt_pred) * 100
print("\n" + "=" * 45)
print("           DECISION TREE RESULTS")
print("=" * 45)
print(" Accuracy :", round(dt_acc, 2), "%")
cm_dt = confusion_matrix(y_test, dt_pred)
sns.heatmap(cm_dt, annot=True, fmt='d', cmap="Greens")
plt.title('Confusion Matrix - Decision Tree')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

dt_report = classification_report(y_test, dt_pred, output_dict=True)
print("-" * 45)
print("\n              DETAILED REPORT")
print("-" * 45)
print("No Diabetes:")
print("Precision :", round(dt_report['0']['precision'] * 100, 2), "%")
print("Recall    :", round(dt_report['0']['recall']    * 100, 2), "%")
print("F1 Score  :", round(dt_report['0']['f1-score']  * 100, 2), "%")
print("\nDiabetes:")
print("Precision :", round(dt_report['1']['precision'] * 100, 2), "%")
print("Recall    :", round(dt_report['1']['recall']    * 100, 2), "%")
print("F1 Score  :", round(dt_report['1']['f1-score']  * 100, 2), "%")
print("=" * 45)

# ---------RANDOM FOREST MODEL------------
rf_model = RandomForestClassifier(n_estimators=100, max_depth=8,random_state=42)
rf_model.fit(x_train, y_train)
rf_pred = rf_model.predict(x_test)
rf_acc = accuracy_score(y_test, rf_pred) * 100

print("\n" + "=" * 45)
print("           RANDOM FOREST RESULTS")
print("=" * 45)
print(" Accuracy :", round(rf_acc, 2), "%")

cm_rf = confusion_matrix(y_test, rf_pred)
sns.heatmap(cm_rf, annot=True, fmt='d', cmap="Oranges")
plt.title('Confusion Matrix - Random Forest')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

rf_report = classification_report(y_test, rf_pred, output_dict=True)
print("-" * 45)
print("\n              DETAILED REPORT")
print("-" * 45)
print("No Diabetes:")
print("Precision :", round(rf_report['0']['precision'] * 100, 2), "%")
print("Recall    :", round(rf_report['0']['recall']    * 100, 2), "%")
print("F1 Score  :", round(rf_report['0']['f1-score']  * 100, 2), "%")
print("\nDiabetes:")
print("Precision :", round(rf_report['1']['precision'] * 100, 2), "%")
print("Recall    :", round(rf_report['1']['recall']    * 100, 2), "%")
print("F1 Score  :", round(rf_report['1']['f1-score']  * 100, 2), "%")
print("=" * 45)

# ----------MODEL COMPARISON-----------
print("\n" + "=" * 45)
print("            MODEL COMPARISON")
print("=" * 45)
print(f"{'Model':<20}{'Accuracy':>15}")
print("-" * 45)
print(f"{'Logistic Regression':<20}{round(acc,2):>13}%")
print(f"{'Decision Tree':<20}{round(dt_acc,2):>13}%")
print(f"{'Random Forest':<20}{round(rf_acc,2):>13}%")
print("=" * 45)

# Feature Importance (Random Forest)
importances = pd.Series(rf_model.feature_importances_, index=x.columns).sort_values(ascending=False)
plt.figure(figsize=(10, 6))
sns.barplot(x=importances.values, y=importances.index,color="steelblue" )
plt.title("Feature Importance - Random Forest")
plt.xlabel("Importance Score")
plt.show()

# ----------NEW PATIENT PREDICTION-------------
print("\n            NEW PATIENT PREDICTION")
print("-" * 45)
new_patient = pd.DataFrame([[1, 100, 62, 25, 67, 22, 0.1, 25]], columns=x.columns)
new_scaled = scaler.transform(new_patient)
print("Glucose=100, BMI=22, Age=25")

lr_result = model.predict(new_scaled)[0]
dt_result = dt_model.predict(new_scaled)[0]
rf_result = rf_model.predict(new_scaled)[0]

print("\nLogistic Regression Prediction :", "DIABETIC" if lr_result == 1 else "NOT DIABETIC")
print("Decision Tree Prediction       :", "DIABETIC" if dt_result == 1 else "NOT DIABETIC")
print("Random Forest Prediction       :", "DIABETIC" if rf_result == 1 else "NOT DIABETIC")

if rf_result == 1:
    print("\nMake and eat healthy food. \nBe active most days.  \nTest your blood sugar often. \nTake medicines as prescribed, even if you feel good......")
else:
    print("\nYOU ARE A HEALTHY PERSON")
print("=" * 45)
```
